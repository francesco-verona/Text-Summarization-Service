```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
import torch

# Scegli il modello che vuoi usare
model_name = "facebook/bart-large-cnn"
summarizer = pipeline("summarization", model=model_name)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
# Carica il modello e il tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name, output_hidden_states=True)
# apri il file in modalità lettura ("r")
with open("articoli_input/testo1.txt", "r", encoding="utf-8") as f:
    text = f.read()


inputs = tokenizer(text, return_tensors="pt", truncation=True,
    max_length=1024)
# Genera il riassunto (in questo caso, gli ID dei token)
summary_ids = model.generate(
    inputs["input_ids"],
    num_beams=4,
    max_length=150,
    min_length=30,
    early_stopping=True
)

# Decodifica gli ID dei token in testo leggibile
summary_ids = summary_ids.sequences  # estrai solo le sequenze di token
summary_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)



# Stampa il riassunto
print("Riassunto generato:", summary_text)
```
